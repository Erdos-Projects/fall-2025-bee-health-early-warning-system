#### Aggregating quarterly data

To do so, let us first define a new column containing quarter information for every sample_month_number

```python
month_quarter_map = {"1": "Q1", "2" : "Q1", "3": "Q1",
 "4": "Q2", "5" : "Q2", "6": "Q2",
 "7": "Q3", "8" : "Q3", "9": "Q3",
 "10": "Q4", "11" : "Q4", "12": "Q4"}

pathogen_new['sample_quarter'] = pathogen_new['sample_month_number'].astype(str).map(month_quarter_map)
last_col_name = pathogen_new.columns[-1]
last_col_data = pathogen_new.pop(last_col_name)
pathogen_new.insert(2, last_col_name, last_col_data)
```

```python
pathogen_new.head()
```

```python
monthly_agg_df = (
    pathogen_new.groupby(["sample_year", "sample_month_number", "state_code","sample_quarter"])
    .agg(
        # Varroa Metrics
        mean_varroa=('varroa_per_100_bees', 'mean'),
        varroa_total_samples=('varroa_per_100_bees', 'count'),
        
        # Nosema Spore Metrics
        mean_spores=('million_spores_per_bee', 'mean'),
        spores_total_samples=('million_spores_per_bee', 'count'),
        
        # Virus Prevalence Metrics
        abpv_prevalence=('abpv', 'mean'),
        abpv_total_samples=('abpv', 'count'),
        cbpv_prevalence=('cbpv', 'mean'),
        cbpv_total_samples=('cbpv', 'count'),
        dwv_prevalence=('dwv', 'mean'),
        dwv_total_samples=('dwv', 'count'),
        iapv_prevalence=('iapv', 'mean'),
        iapv_total_samples=('iapv', 'count'),
        kbv_prevalence=('kbv', 'mean'),
        kbv_total_samples=('kbv', 'count'),
        lsv2_prevalence=('lsv2', 'mean'),
        lsv2_total_samples=('lsv2', 'count'),
        
        unique_counties_sampled=('sampling_county', 'nunique') 
    )
    .reset_index()
)
```

```python
monthly_agg_df.head(50)
monthly_agg_df.to_csv("../data/processed/pathogen_data_monthly_aggregated_2015_onwards.csv", index=False)
cp_monthly_agg_df = monthly_agg_df.copy(deep=True)
```

```python
# Define pathogen lists
pathogens = ['varroa', 'spores', 'abpv', 'cbpv', 'dwv', 'iapv', 'kbv', 'lsv2']
mean_cols = ['mean_varroa', 'mean_spores']
prev_cols = ['abpv_prevalence', 'cbpv_prevalence', 'dwv_prevalence', 'iapv_prevalence', 'kbv_prevalence', 'lsv2_prevalence']
all_measure_cols = mean_cols + prev_cols

cp_monthly_agg_df[all_measure_cols] = cp_monthly_agg_df[all_measure_cols].fillna(0)

# Calculate weighted monthly values
for pathogen in pathogens:
    if pathogen in ['varroa', 'spores']:
        measure_col = f'mean_{pathogen}'
        total_col = f'{pathogen}_total_samples'
    else:
        measure_col = f'{pathogen}_prevalence'
        total_col = f'{pathogen}_total_samples'
    
    weighted_col = f'{pathogen}_weighted_sum' 
    cp_monthly_agg_df[weighted_col] = cp_monthly_agg_df[measure_col].fillna(0) * cp_monthly_agg_df[total_col]

# Sort data for correct 'last' aggregation
cp_monthly_agg_df = cp_monthly_agg_df.sort_values(by=['sample_year', 'state_code', 'sample_month_number'])

cp_monthly_agg_df.head()
```

```python
# Define grouping and aggregations
group_keys = ['sample_year', 'sample_quarter', 'state_code']
agg_dict = {}

for pathogen in pathogens:
    if pathogen in ['varroa', 'spores']:
        measure_col = f'mean_{pathogen}'
        total_col = f'{pathogen}_total_samples'
    else:
        measure_col = f'{pathogen}_prevalence'
        total_col = f'{pathogen}_total_samples'
    
    weighted_col = f'{pathogen}_weighted_sum'

    # Add aggregations to dict
    agg_dict[f'{total_col}_q'] = (total_col, 'sum')
    agg_dict[f'{weighted_col}_q'] = (weighted_col, 'sum') # (was f'{quarterly_avg}_q')
    agg_dict[f'worst_month_{measure_col}'] = (measure_col, 'max')
    agg_dict[f'last_month_{measure_col}'] = (measure_col, 'last')

# Add other aggregations
agg_dict['unique_counties_sampled_q'] = ('unique_counties_sampled', 'sum')
agg_dict['last_month_in_quarter'] = ('sample_month_number', 'max')

# Perform aggregation
quarterly_df = cp_monthly_agg_df.groupby(group_keys).agg(**agg_dict).reset_index()

# Post-Aggregation Calculations (Quarterly Means)
final_columns_order = list(group_keys) # Start with grouping keys

# Lists to hold new column names for ordering
quarterly_mean_cols = []
quarterly_total_cols = []
worst_month_cols = []
last_month_cols = []

for pathogen in pathogens:
    if pathogen in ['varroa', 'spores']:
        measure_col_name = f'mean_{pathogen}'
        q_mean_col_name = f'mean_{pathogen}_q'
    else:
        measure_col_name = f'{pathogen}_prevalence'
        q_mean_col_name = f'{pathogen}_prevalence_q'

    total_col_q_name = f'{pathogen}_total_samples_q'
    
    # --- FIX 2 ---
    # Use the name that was created in the aggregation dict
    # (was '_weighted_sum_q')
    weighted_col_q_name = f'{pathogen}_weighted_sum_q' 
    worst_col_name = f'worst_month_{measure_col_name}'
    last_col_name = f'last_month_{measure_col_name}'

    # Calculate quarterly mean (weighted average)
    quarterly_df[q_mean_col_name] = quarterly_df[weighted_col_q_name] / quarterly_df[total_col_q_name]
    
    # Handle division by zero (replace NaN with 0)
    quarterly_df[q_mean_col_name] = quarterly_df[q_mean_col_name].fillna(0)

    # Add to ordered lists
    quarterly_mean_cols.append(q_mean_col_name)
    quarterly_total_cols.append(total_col_q_name)
    worst_month_cols.append(worst_col_name)
    last_month_cols.append(last_col_name)

# Define the final
final_columns_order.extend(quarterly_mean_cols)
final_columns_order.extend(quarterly_total_cols)
final_columns_order.extend(worst_month_cols)
final_columns_order.extend(last_month_cols)
final_columns_order.append('unique_counties_sampled_q')
final_columns_order.append('last_month_in_quarter')

# Keep only the relevant columns in the defined order
quarterly_df_final = quarterly_df[final_columns_order]

# Save the aggregated data
output_file_path = 'pathogen_data_quarterly_aggregated.csv'
quarterly_df_final.to_csv(output_file_path, index=False)

# Print head and info of the new dataframe
print(f"Aggregated data saved to: {output_file_path}")
print("\nQuarterly Aggregated DataFrame Head:")
print(quarterly_df_final.head())
print("\nQuarterly Aggregated DataFrame Info:")
quarterly_df_final.info()
```
